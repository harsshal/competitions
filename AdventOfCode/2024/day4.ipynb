```python
from itertools import product
from collections import Counter
```

```python
with open("/home/harshal/projects/competitions/AdventOfCode/2024/day4.in") as f:
    # readlines doesnt remove '\n' from the end
    data = f.read().strip().split("\n")

data
```

```python
directions = set(product((-1,0,1),(-1,0,1))) - {(0,0)}
#print(directions)

total = 0
for row_index, row in enumerate(data):
    for col_index, col in enumerate(row):
        for dir in directions:
            # can add padding of '###' at the start and the end to deal with index error as well

            # try:
            #     if [data[row_index + i * dir[0]][col_index + i * dir[1]] for i in range(4)] == ["X","M","A","S"]:
            #         total+=1
            # except IndexError:
            #     continue

            # Check bounds
            if not (0 <= row_index + 3 * dir[0] < len(data) and
                    0 <= col_index + 3 * dir[1] < len(row)):
                continue
            # Check the pattern
            pattern = [data[row_index + i * dir[0]][col_index + i * dir[1]] for i in range(4)]
            if pattern == ["X", "M", "A", "S"]:
                total += 1

print(total)
```

```python
total = 0
for row_index, row in enumerate(data):
    for col_index, col in enumerate(row):

        if col != "A":
            continue

        # Check bounds
        if not (1 <= row_index < len(data)-1 and
                1 <= col_index < len(row)-1):
            continue

        ms_set = [ data[row_index -1][col_index -1],
                  data[row_index -1][col_index +1],
                  data[row_index +1][col_index -1],
                  data[row_index +1][col_index +1]
                ]

        if (Counter(ms_set) == {"M":2, "S":2}) and ms_set[0] != ms_set[3]:
            #print(row_index, col_index, ms_set)
            total += 1

print(total)
```

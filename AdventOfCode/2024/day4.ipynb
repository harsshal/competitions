```python
from itertools import product
```

```python
directions = set(product((-1,0,1),(-1,0,1))) - {(0,0)}

total = 0
for row_index, row in enumerate(data):
    for col_index, col in enumerate(row):
        for dir in directions:
            try:
                if [data[row_index + i * dir[0]][col_index + i * dir[1]] for i in range(4)] == ["X","M","A","S"]:
                    total+=1
            except IndexError:
                continue

print(total)
```

```python
# from https://github.com/fuglede/adventofcode/blob/master/2024/day04/solutions.py

from collections import defaultdict
from itertools import product

with open("/home/harshal/projects/competitions/AdventOfCode/2024/day4.in") as f:
    ls = f.read().strip().split("\n")


boardz = defaultdict(str)
print(boardz)
boardz |= {i + 1j * j: x for i, l in enumerate(ls) for j, x in enumerate(l)}
print(boardz)
octdir = {i + 1j * j for (i, j) in set(product((-1, 0, 1), (-1, 0, 1))) - {(0, 0)}}
print(octdir)

# Part 1
print(
    sum(
        [boardz[z + i * dz] for i in range(4)] == ["X", "M", "A", "S"]
        for z in list(boardz.keys())
        for dz in octdir
    )
)


# Part 2
res = 0
for z in list(boardz.keys()):
    if boardz[z] == "A":
        corners = [
            boardz[z + 1 + 1j],
            boardz[z + 1 - 1j],
            boardz[z - 1 - 1j],
            boardz[z - 1 + 1j],
        ]
        if (
            corners.count("M") == 2
            and corners.count("S") == 2
            and boardz[z - 1 - 1j] != boardz[z + 1 + 1j]
        ):
            res += 1
print(res)
```

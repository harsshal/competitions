```python
def find_safe(ser):
    #print(ser)
    origser = ser.diff().dropna().astype(int)
    #print(ser)
    if all(origser.isin([1,2,3])) or all(origser.isin([-1,-2,-3])):
        return 1
    
    for i in range(len(ser)):
        origser = ser.drop(ser.index[i])
        origser = origser.diff().dropna().astype(int)
        if all(origser.isin([1,2,3])) or all(origser.isin([-1,-2,-3])):
            return 1
    
    return 0

data_in.apply(find_safe,axis=1).sum()
```

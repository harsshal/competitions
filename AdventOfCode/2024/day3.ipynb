```python
with open("/home/harshal/projects/competitions/AdventOfCode/2024/day3.in") as f:
    data = f.read()

data
```

```python
import re
```

```python
total = 0
for num1, num2 in re.findall("mul\((\d+),(\d+)\)", data):
    #print(num1,num2)
    total += int(num1) * int(num2)
print(total)
```

```python
total = 0
enabled = True
for cmd, num1, num2 in re.findall("(mul\((\d+),(\d+)\)|do\(\)|don't\(\))", data):
    print(cmd,num1,num2)
    if cmd == "don't()":
        enabled = False
    elif cmd == "do()":
        enabled = True
    else:
        if enabled:
            total += int(num1) * int(num2)
print(total)
```

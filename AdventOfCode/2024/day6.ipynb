```python
with open("/home/harshal/projects/competitions/AdventOfCode/2024/day6.in") as f:
    data = f.read().strip().split("\n")
data
```

```python
start_location = [(r, c) for r, row in enumerate(data) for c, char in enumerate(row) if char == '^'][0]
start_location
```

```python
directions = [(-1,0),(0,1),(1,0),(0,-1)]
```

```python
def walk(data, location, direction):
    new_data = []
    for str_index, str in enumerate(data):
        if str_index != location[0]:
            new_data.append(str)
        else:
            new_data.append(str[:location[1]]+'X'+str[location[1]+1:])
    data = new_data
    
    next_location = location + direction

    if not (0 <= next_location[0] < len(data)) or not (0 <= next_location[1] < len(data[next_location[0]])):
        return len([[(r, c) for r, row in enumerate(data) for c, char in enumerate(row) if char == 'X']])
    else:
        if data[next_location[0]][next_location[1]] == '#':
            next_direction = (directions.index(direction) + 1) % 4
            walk(data,location + next_direction, next_direction)
        else:
            walk(data,location+direction, direction)

walk(data, start_location, directions[0])
```

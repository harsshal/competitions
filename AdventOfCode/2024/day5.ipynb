```python
rules = data[:data.index('')]
clean_rules = list(map(lambda x:x.split('|'),rules))
updates = data[data.index('')+1:]
clean_updates = list(map(lambda x:x.split(','),updates))


for rule in clean_rules:
    incorrect_updates = []
    for update in clean_updates:
        if rule[0] in update and rule[1] in update and update.index(rule[0]) > update.index(rule[1]):
           incorrect_updates.append(update)

incorrect_updates = [list(x) for x in set(tuple(x) for x in incorrect_updates)]

total = 0
for update in incorrect_updates:
    while True:
        rerun_rules = False
        for rule in clean_rules:
            if rule[0] in update and rule[1] in update and update.index(rule[0]) > update.index(rule[1]):
                index_0, index_1 = update.index(rule[0]), update.index(rule[1])
                update[index_1], update[index_0] = update[index_0], update[index_1]
                rerun_rules = True

        if not rerun_rules:
            break

    total += int(update[len(update)//2])

print(total)
```

```python
# From : https://github.com/fuglede/adventofcode/blob/master/2024/day05/solutions.py

import networkx as nx

with open("/home/harshal/projects/competitions/AdventOfCode/2024/day5.in") as f:
    gs = f.read().strip().split("\n\n")

rules = [tuple(map(int, l.split("|"))) for l in gs[0].split("\n")]
updates = [tuple(map(int, l.split(","))) for l in gs[1].split("\n")]

unsorted_updates = {
    update
    for update in updates
    if any(
        a in update and b in update and update.index(a) > update.index(b)
        for a, b in rules
    )
}

print(unsorted_updates)

# Part 1
print(
    sum(
        update[len(update) // 2] for update in updates if update not in unsorted_updates
    )
)

# Part 2
sorted_updates = [
    list(
        nx.topological_sort(
            nx.DiGraph((a, b) for a, b in rules if a in update and b in update)
        )
    )
    for update in unsorted_updates
]

print(sum(update[len(update) // 2] for update in sorted_updates))
```

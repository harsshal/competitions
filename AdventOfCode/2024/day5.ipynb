```python
import networkx as nx
```

```python
with open("/home/harshal/projects/competitions/AdventOfCode/2024/day5.in") as f:
    # readlines doesnt remove '\n' from the end
    data = f.read().strip().split("\n")

data
```

```python
rules = data[:data.index('')]
clean_rules = list(map(lambda x:x.split('|'),rules))
updates = data[data.index('')+1:]
clean_updates = list(map(lambda x:x.split(','),updates))


for rule in clean_rules:
    filtered_updates = []
    for update in clean_updates:
        if rule[0] in update and rule[1] in update and update.index(rule[0]) > update.index(rule[1]):
            continue
        filtered_updates.append(update)

    clean_updates = filtered_updates

# // is integer division
sum(list(map(lambda x:int(x[len(x)//2]),clean_updates)))
```

```python
rules = data[:data.index('')]
clean_rules = list(map(lambda x:tuple(map(int,x.split('|'))),rules))
updates = data[data.index('')+1:]
clean_updates = list(map(lambda x:tuple(map(int,x.split(','))),updates))


incorrect_updates = []
for update in clean_updates:
    for rule in clean_rules:    
        if rule[0] in update and rule[1] in update and update.index(rule[0]) > update.index(rule[1]):
           incorrect_updates.append(update)
           break

#print(len(clean_rules), len(clean_updates), len(incorrect_updates))

total = 0
for update in incorrect_updates:
    #print(update)
    DG = nx.DiGraph((int(a), int(b)) for a,b in clean_rules if a in update and b in update)
    #print(DG)
    sorted_update = list(nx.topological_sort(DG))
    #print(sorted_update)
    #print(total, len(sorted_update), len(sorted_update) // 2)
    total += sorted_update[len(sorted_update) // 2]

print(total)
```
